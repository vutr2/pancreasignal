```python
# Fixed: TITLE_ABS:"CA19-9" with NO space after the colon (with a space -> 0 hits, silently)
import requests

query = '(TITLE_ABS:"CA19-9" OR TITLE_ABS:"CA 19-9") AND TITLE_ABS:"pancreatic" AND OPEN_ACCESS:y'
# Narrower version (uncomment to use): reviews since 2019 about false positives / limitations
# query += ' AND PUB_TYPE:"review" AND PUB_YEAR:[2019 TO 2026] AND (TITLE_ABS:"false positive" OR TITLE_ABS:"limitation")'

r = requests.get("https://www.ebi.ac.uk/europepmc/webservices/rest/search", params={
    "query": query, "format": "json", "resultType": "core", "pageSize": 25,
    "sort": "CITED desc",                 # most-cited first
})
r.raise_for_status()                      # an HTTP error stops here instead of failing silently
data = r.json()
print("hitCount:", data["hitCount"])      # ALWAYS print this first: 0 means the query is wrong

for a in data["resultList"]["result"]:
    print(a.get("pmcid"), a.get("pubYear"), a.get("license"), a["title"][:90])
```
